```python
# Function that Changes its Input
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

A function that doesn't change it's data is likely preferred. A helper function might do something wrong, it could be necessary to compare to the original data you were given. If you change the input, you are erasing what you already know.

12) Choosing Data Structures
A. A sequence of ages where order matters.
A list/array would be best here. The items are hard-ordered, and they cannot be changed without a complete reassignment. 

B. A fixed latitude/longitude coordinate.
A tuple would be best, it's immutable and has an order (N/S, E/W)

C. Subject ID → treatment group.
A dictionary would map these values. Given a key for the subject ID, it could store who has what treatment.

D. Unique treatment groups.
A set would be best. It would store each unique treatment.

E. A collection that will be repeatedly modified.
A set would be best, since it has no distinct order and would allow for constant modification.

```python
# Data Science Application
participants = [
    {"id": 101, "age": 24, "group": "A", "score": 81},
    {"id": 102, "age": 31, "group": "B", "score": 94},
    {"id": 103, "age": 27, "group": "A", "score": 88},
    {"id": 104, "age": 19, "group": "B", "score": 72},
    {"id": 105, "age": 42, "group": "A", "score": 91},
    {"id": 106, "age": 35, "group": "B", "score": 85},
]
```

```python
ids = [part["id"] for part in participants]
scores = [part["score"] for part in participants]
high_scores = [part["score"] for part in participants if part["score"] >= 85]
groups = {part["group"] for part in participants}
score_by_id = dict([part["id"], part["score"]] for part in participants)

scores
```

```python
summary = {
    "n": len(ids),
    "groups": groups,
    "mean_score": sum(scores) / len(ids),
    "max_score": max(high_scores),
    "high_performers": dict([part["id"], part["score"]] for part in participants if part["score"] >= 85)
}
```

```python
# Loops vs Comprehensions
# comprehension
scores = [part["score"] for part in participants]

#vs loop
scores = []
for part in participants:
    scores.append(part["score"])
```

The comprehension takes about 30 ms less to calculate, and is about as easy to read as the loop. It doesn't require the initialization of the variable, either.
