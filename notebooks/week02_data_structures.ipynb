```python
# Function that Changes its Input
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

A function that doesn't change it's data is likely preferred. A helper function might do something wrong, it could be necessary to compare to the original data you were given. If you change the input, you are erasing what you already know.

```python
# Shallow Copying
x = [[1, 2], [3, 4]]
y = x.copy()

print(x is y)
print(x[0] is y[0])
```

```python
x[0].append(99)

print(x)
print(y)
```

x is a list of lists. Because y is copying x, the list in y changes as well, even if they're defined as separate objects.
