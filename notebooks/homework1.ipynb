3) Joining Tables
The first two tables I will join are the enrollments and the students. This makes the most sense to me, since I will only be adding the information from the students to each enrollment. I will join based on Student ID from both tables.

```python
# merge the tables based on the id, expecting a one-to-one relationship
enrollDetail = enrollments.merge(
    students,
    on = "student_id",
    how = "outer",
    validate="many_to_one"
)
enrollDetail.head()
```

The resulting dataset adds the student details to the enrollment. Upon inspection, there is one row more than in the original enrollments dataset. One of the students have no enrollments, and I chose an outer join, meaning all keys in both datasets are present, even if they aren't present in the other. This is also mirrored, since there is also an enrollment that presents a student ID not present in the students data.

I validated with a many to one relationship, since students can enroll in several courses, but each enrollment can only have one student. This presented no issues. From here, we can now join the courses table.

```python
studentClassDetail = enrollDetail.merge(
    courses,
    on = "course_id",
    how = "outer",
    validate="many_to_many",
    indicator=True
)

studentClassDetail.head()
```

Once again, I validated a many-to-many relationship, since students can have multiple courses, and courses can have multiple students. The indicator shows that most rows came from both data sets, meaning most have matches, but two rows contain a course that does not have information from the courses dataset.No new or duplicate rows were created during this merge, which indicates that there are no courses in the courses dataset that have 0 enrollments.

```python
#reshape the data
scores_long = studentClassDetail.melt(
    id_vars=["enrollment_id", "student_id", "course_id"],
    value_vars=["midterm_score", "final_score", "project_score"],
    var_name="assessment",
    value_name="score"
)

scores_long.sample(n=7)
```

I took a random sample of the data to show the change. I changed the midterm_score, final_score, and project_score into one "assessment" column which shows which of the 3 is being described. The enrollment ID, student ID, and course ID are kept to make sense of the assessments.

Replacing all missing values with zeros would have drastic consequences. Many of the IDs are strings and non-numeric. Most importantly, for the scores missing, a zero would indicate a failure of completion instead of a true missing value. This would not accurately reflect the student's performance.
