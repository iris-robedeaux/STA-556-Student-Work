```python
# Data Science Test
scores = [
    72,
    81,
    91,
    68,
    88,
    95,
    77,
    84
]

sum(scores) / len(scores)
```

```python
import numpy as np

scores = np.array(
    scores
)

scores.mean()
scores.std()
```

```python
# Create a plot
import matplotlib.pyplot as plt

plt.hist(
    scores
)
plt.xlabel(
    "Score"
)
plt.ylabel(
    "Frequency"
)
plt.title(
    "Example Score Distribution"
)
plt.show()
```

```python
# Save the plot
plt.hist(
    scores
)

plt.savefig(
    "/workspaces/STA-556-Student-Work/figures/scores.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
