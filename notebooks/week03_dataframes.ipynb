The .loc command is cleaner since it's cleaner to read. The loc command more naturally selects on conditions, but subsetting a df on a condition is awkward when subsetting the traditional way. 1. DataFrames
What does a pandas DataFrame provide that a list of dictionaries does not?
pandas can list column titles. Dictionaries can list keys, their data types, but it cannot store a name for a set of keys.

2. Series
What is the difference between:
df["age"]
and:
df[["age"]]?
The first creates a series and the second creates a dataframe.

3. .loc vs. .iloc
Explain the difference in one sentence.
.loc uses values to subset data, but iloc uses direct indicies for rows and columns to subset.

4. Boolean filtering
What does:
df["score"] >= 85
produce before it is used to subset a DataFrame?
Before it's subset, a boolean list of whether each row follows the condition is created.

5. Analysis populations
Why is a pandas filter more than simply a programming operation?
A pandas filter means you can filter on values, narrowing data down easier to subsets you want to use to analyze. It allows for the ease of data analysis.

6. Validation
Why should you inspect a filtered dataset even if no Python error occurred?
It's vital to ensure that the command works. For example, > or >= can be the difference between including or excluding a valid observation. It's necessary to at least spot check and ensure the code is doing what you want it to.

7. Reproducibility
How can explicit filtering criteria make statistical analyses easier to reproduce?
Using explicit criteria means any onlooker can understand what your code is doing. If someone approaches your code without being able to understand it, that could be problematic in a job scenario, especially were a supervisor might be overlooking your finished work.
